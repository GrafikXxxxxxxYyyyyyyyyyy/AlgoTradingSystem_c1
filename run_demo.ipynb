```python
import sys
import asyncio
import logging
from xgboost import XGBRegressor
from src.trading.demo_trader import DemoTrader

logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(name)s - %(levelname)s - %(message)s",
    handlers=[logging.StreamHandler(sys.stdout)],
)

mid_model = XGBRegressor()
mid_model._estimator_type = "regressor"
mid_model.load_model("models/model_mid.json")

spread_model = XGBRegressor()
spread_model._estimator_type = "regressor"
spread_model.load_model("models/model_spread.json")

trader = DemoTrader(
    mid_model=mid_model,
    spread_model=spread_model,
    alpha=2.15,
    beta=1.48,
    gamma=0.0,
    epsilon=0.024,
    max_position=0.01,
    position_qty=0.002,
    log_dir="./runs/demo",
)

# В Jupyter можно: await trader.run(symbol="BTCUSDC", grid_resolution_ms=5000)
# В скрипте: asyncio.run(trader.run(symbol="BTCUSDC", grid_resolution_ms=5000))
await trader.run(symbol="BTCUSDC", grid_resolution_ms=5000)
```
